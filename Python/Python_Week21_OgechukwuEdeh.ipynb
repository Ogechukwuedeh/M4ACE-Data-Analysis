```python
# USSD app for Access bank
# Variables
bank = "Access Bank"
ussd_code = "*901#"
balance = 50000

# Function to check balance
def check_balance():
    print(f"Your balance is ₦{balance}")

# Function to transfer money
def transfer():
    account = input("Enter account number: ")
    amount = input("Enter amount: ₦")
    print(f"₦{amount} transfered successfully to {account}")

# Function to buy airtime
def buy_airtime():
    amount = input("Enter airtime amount: ₦")
    print(f"You have successfully purchased ₦{amount} airtime. ")

# Main USSD function
def ussd():
    code = input("Dail USSD code: ")

    if code == ussd_code:
        while True:
            print("\nWelcome to", bank)
            print("1. Check Balance")
            print("2. Transfer")
            print("3. Buy Airtime")
            print("4. Exit")

            choice = input("Select an option: ")

            if choice == "1":
                check_balance()

            elif choice == "2":
                transfer()

            elif choice == "3":
                buy_airtime()

            elif choice == "4":
                print("Thank you for banking with us.")
                break

        else:
            print("Invalid option. Try again.")

    else:
        print("Invalid USSD code.")

# Run the program
ussd()
            
```
